```python
# Step 1: Change / to -
df['order_date'] = df['order_date'].str.replace('/', '-', regex=False)

# Step 2: Convert different date formats into datetime
df['order_date'] = pd.to_datetime(
    df['order_date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)

# Step 3: Convert everything to YYYY-MM-DD
df['order_date'] = df['order_date'].dt.strftime('%Y-%m-%d')

```

## Category wise sales

```python
df.groupby('product_category')['sales_amount'].sum().plot(kind='bar')

plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales Amount')
plt.show()

```

## Monthly Sales

```python
df['order_date'] = pd.to_datetime(df['order_date'])

monthly_sales = df.groupby(
    df['order_date'].dt.to_period('M')
)['sales_amount'].sum()
```

```python
monthly_sales.plot(kind='line')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.show()
```

## Profit Distribution

```python
df['profit'].plot(kind='hist', bins=20)

plt.title('Profit Distribution')
plt.xlabel('Profit')
plt.ylabel('Frequency')
plt.show()
```

## Discount vs Profit

```python
plt.scatter(df['discount_pct'], df['profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```
